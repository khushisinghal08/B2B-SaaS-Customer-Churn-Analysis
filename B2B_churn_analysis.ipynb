```python
Q1 = df['Monthly_Recurring_Revenue'].quantile(0.25)
Q3 = df['Monthly_Recurring_Revenue'].quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + (1.5 * IQR)

print("Lower Quartile:", Q1)
print("Upper Quartile:", Q3)
print("IQR:", IQR)
print("Upper Outlier Limit:", upper_limit)

outliers = df[df['Monthly_Recurring_Revenue'] > upper_limit]

print("Potential Revenue Outliers:", len(outliers))
```

###Data Analysis & Business Insights.

```python
#Overall Churn Analysis

churn_counts = df['Is_Churned'].value_counts()
churn_rate = df['Is_Churned'].mean() * 100

print("Churn Distribution:")
print(churn_counts)

print("Overall Churn Rate:", round(churn_rate, 2), "%")
```

```python
#Churn by Plan Tier

plan_churn = df.groupby('Plan_Tier')['Is_Churned'].agg(
    ['count', 'sum', 'mean']
)

plan_churn['Churn_Rate'] = plan_churn['mean'] * 100

print(plan_churn)
```

```python
#churn by feature adoption

feature_churn = df.groupby('Is_Churned')['Feature_Adoption_Rate'].mean()

print(feature_churn)
```

```python
#churn by seat utilization

seat_churn = df.groupby('Is_Churned')['Seat_Utilization'].mean()

print(seat_churn)
```

```python
#churn by support ticket
support_churn = df.groupby('Is_Churned')['Support_Tickets_30d'].mean()

print(support_churn)
```

```python
# Revenue by subscription plan
revenue_by_plan = df.groupby('Plan_Tier')['Monthly_Recurring_Revenue'].agg(
    ['count', 'sum', 'mean']
)

print(revenue_by_plan)
```
